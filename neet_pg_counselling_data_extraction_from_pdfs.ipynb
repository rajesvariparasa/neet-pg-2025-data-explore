```python
#@title Helper Functions

def clean(text):
    if text is None:
        return ""
    return re.sub(r"\s+", " ", str(text)).strip()

def normalize_row(cells):
    """
    Pads / trims row to exactly 6 columns
    """
    cells = [clean(c) for c in cells]
    if len(cells) < len(COLUMNS):
        cells += [""] * (len(COLUMNS) - len(cells))
    return cells[:len(COLUMNS)]

def page_chunks(total, size):
    for start in range(1, total + 1, size):
        end = min(start + size - 1, total)
        yield f"{start}-{end}"

def main():
    # Count pages
    with pdfplumber.open(PDF_PATH) as pdf:
        total_pages = len(pdf.pages)

    all_rows = []

    for page_range in page_chunks(total_pages, CHUNK_SIZE):
        print(f"Processing pages {page_range}")
        tables = camelot.read_pdf(
            PDF_PATH,
            pages=page_range,
            flavor="lattice",
            strip_text="\n"
        )

        for table in tables:
            df = table.df

            for _, row in df.iterrows():
                cells = normalize_row(row.tolist())

                # Skip ONLY true header rows
                joined = " ".join(cells).lower()
                if "sno" in joined and "rank" in joined and "institute" in joined:
                    continue

                all_rows.append(dict(zip(COLUMNS, cells)))

    final_df = pd.DataFrame(all_rows, columns=COLUMNS)
    display(final_df)
    final_df.to_csv(OUTPUT_CSV, index=False)

    print(f"\n✅ Extracted {len(final_df)} rows to {OUTPUT_CSV}")
```

# Clean up

```python
import pandas as pd
df_r1 = pd.read_csv("r1_neet_pg_2025.csv")
df_r2 = pd.read_csv("r2_neet_pg_2025.csv")
df_r3 = pd.read_csv("r3_neet_pg_2025.csv")
```

```python
#combine rank and group to create unique ID
df_r1 = df_r1.rename(columns={"R1_Rank": "Rank", "R1_Course":"Course" ,"R1_Institute": "Institute",
                              "R1_Allot_Quota":"Allot_Quota","R1_Allot_Cat": "Allot_Cat" ,"R1_Cand_Cat":"Cand_Cat",
                              "R1_Remarks": "Remarks"})
df_r1 = df_r1.drop(columns=["sno"])
df_r1["Round"] = 1
df_r1
```

```python
df_r2 = df_r2.rename(columns={"R1_Rank": "Rank", "R2_Course":"Course" ,"R2_Institute": "Institute",
                              "R2_Allot_Quota":"Allot_Quota","R2_Allot_Cat": "Allot_Cat" ,"R2_Cand_Cat":"Cand_Cat",
                             "R2_Remarks": "Remarks"})
df_r2 = df_r2.drop(columns=["R1_Institute", "R1_Course", "R1_Remarks", "R1_Allot_Quota","R2_Opt_No" ])
df_r2 = df_r2[df_r2.Rank!= "-"]
df_r2["Round"] = 2
df_r2
```

```python
df_r3 = df_r3.rename(columns={"R1_Rank": "Rank", "R3_Course": "Course","R3_Institute": "Institute",
                              "R3_Allot_Quota":"Allot_Quota","R3_Allot_Cat": "Allot_Cat" ,"R3_Cand_Cat":"Cand_Cat",
                               "R3_Remarks": "Remarks"})
df_r3 = df_r3.drop(columns=["R1_Institute", "R2_Institute", "R1_Allot_Quota",	"R2_Allot_Quota",
                            "R1_Course", "R2_CourseID", "R1_Remarks", "R2_Remarks" ,
                            "R3_Opt_No"
                            ])
df_r3 = df_r3[df_r3.Rank!= "-"]
df_r3["Round"] = 3
df_r3
```

```python
df_r1.shape, df_r2.shape, df_r3.shape
```

```python
#@title Standardise names of institutes, groups, and courses across the three rounds

#concatenate all the dfs
df = pd.concat([df_r1, df_r2, df_r3], ignore_index=True)

#from quick manual skim of the data it seems that non-uniform spacing throughout
#the names is causing same insts/courses to show up as unique entries.
#This will cause issues when we want to group data by inst/course

print("Unique values before removing spaces")

print("Institute names: ", len(df["Institute"].unique()))
print("Alloted Quota:", len(df["Allot_Quota"].unique()))
print("Course names: ",len(df["Course"].unique()))

df["Institute"] = df["Institute"].str.replace(' ', '')
df["Allot_Quota"] = df["Allot_Quota"].str.replace(' ', '')
df["Course"] = df["Course"].str.replace(' ', '')

print("\nUnique values after removing spaces")

print("Institute names: ",len(df["Institute"].unique()))
print("Course names: ", len(df["Course"].unique()))
print("Alloted Quota:", len(df["Allot_Quota"].unique()))

# #now uniformly add spaces to institute names/addresses after each comma in the string
df["Institute"] = df["Institute"].str.replace(',', ', ')
df.head(5)
```

```python
df = pd.read_csv('neet_pg_2025_results_long_format_v2.tsv', sep='\t')
```

```python
#@title Identifying final round of allotment for each candidate

import numpy as np
# finds the latest round of all rounds the candidate participated in
latest = (
    df
    .groupby("Rank", as_index=False)["Round"]
    .max()
    .rename(columns={"Round": "latest_round"})
)

df = df.merge(latest, on="Rank", how="left")

# If latest round value for that candidate matches the round in that row, assign final assignment status to that row
df["Round_assignment_status"] = np.where(
   (df["Round"] == df["latest_round"]),
    "Final assignment",
    ""
)
df=df.drop(columns=["latest_round"])
df.to_csv("neet_pg_2025_results_long_format_v3.tsv", sep="\t")
df.head(20)
```
